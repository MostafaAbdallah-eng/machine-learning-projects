Clustering

```python
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
x,y=make_blobs(n_samples=500,n_features=2,centers=4,random_state=1)
```

```python
print(x)
```

```python
print(y)
```

```python
fig, ax1=plt.subplots(1)
ax1.scatter(x[:,0],x[:,1],marker='o',s=8)
plt.show()
```

```python
color=["red","pink","yellow","green"]
fig,ax1=plt.subplots(1)

for i in range(4):
  ax1.scatter(x[y==i,0],x[y==i,1],marker='o',s=8,c=color[i])

plt.show()
```

```python
n_clusters=8
cluster1=KMeans(n_clusters=n_clusters,random_state=8).fit(x)
```

```python
y_pred1=cluster1.labels_
print(y_pred1)
```

```python
centers=cluster1.cluster_centers_
print(centers)
```

```python
color=["red","pink","yellow","green","orange","blue","gray","brown"]
fig,ax1=plt.subplots(1)

for i in range(8):
  ax1.scatter(x[y_pred1==i,0],x[y_pred1==i,1],marker='o',s=8,c=color[i])

ax1.scatter(centers[:,0],centers[:,1],marker="x",s=15,c="black")

plt.show()
```

```python
nullnew=candf.isnull().sum()
print(nullnew)
```

```python
candf=candf.drop(['competitorname'],axis=1)
```

```python
print(candf)
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import KFold
import tensorflow as tf

kfold=KFold(n_splits=8,shuffle=True)

fold_no=1
x=candf.to_numpy()

for train,test in kfold.split(x):
  xtrain,xtest=x[train],x[test]
  dftrain=pd.DataFrame(xtrain)
  dftest=pd.DataFrame(xtest)

  filepathtrain="/content/drive/MyDrive/Round 4 AI Lab/Candy Kfolds/candftrain"+str(fold_no)+".csv"
  filepathtest="/content/drive/MyDrive/Round 4 AI Lab/Candy Kfolds/candftest"+str(fold_no)+".csv"
  dftrain.to_csv(filepathtrain)
  dftest.to_csv(filepathtest)


  print(fold_no)
  print(dftrain)
  print(dftest)
  fold_no=fold_no+1
```

```python
candfkf1=pd.read_csv('/content/drive/MyDrive/Round 4 AI Lab/Candy Kfolds/candftrain1.csv')
```

```python
targetkf1=candfkf1['11']
candfkf1=candfkf1.drop(['Unnamed: 0','11'],axis=1)

print(targetkf1)
print(candfkf1)
```

```python
from keras.models import Sequential
from keras.layers import Dense
import tensorflow as tf
```

```python
def myregressionMod():
  model=Sequential()
  model.add(Dense(11,input_dim=11,activation='relu'))
  model.add(Dense(9,activation='relu'))
  model.add(Dense(6,activation='relu'))
  model.add(Dense(3,activation='relu'))
  model.add(Dense(1,activation='relu'))
  model.compile(loss='mean_squared_error',optimizer='adam',metrics=['mean_squared_error'])
  return model
```

```python
def setlearningRate(epoch):
  return 0.1
```

```python
Mymodel=myregressionMod()
filepath='/content/drive/MyDrive/Round 4 AI Lab/bestRegkf1.h5'
checkpoint=tf.keras.callbacks.ModelCheckpoint(filepath,monitor='loss',verbose=2,save_best_only=True,mode='min')
Lrcall=tf.keras.callbacks.LearningRateScheduler(setlearningRate)
hist=Mymodel.fit(candfkf1,targetkf1,epochs=200,batch_size=30,callbacks=[checkpoint,Lrcall])
```
